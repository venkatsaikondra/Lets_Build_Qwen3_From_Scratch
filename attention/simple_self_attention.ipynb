```python
import torch
inputs=torch.tensor(
    [
        [0.43,0.15,0.89],#your (x^1)
        [0.55,0.87,0.66],#journey (x^2)
        [0.57,0.85,0.64],#starts (x^3)
        [0.22,0.58,0.33],#with (x^4)
        [0.77,0.25,0.10],#one (x^5)
        [0.05,0.80,0.55]# step (x^6)
    ]
)
```

```python
input_query=inputs[1]
input_query
```

```python
input_1=inputs[0]
input_1
```

```python
torch.dot(input_query,input_1)
```

```python
res=0
for idx,ele in enumerate(inputs[0]):
    res+=inputs[0][idx]*input_query[idx]

print(res)
```

```python
query=inputs[1]
attention_scores_2=torch.empty(inputs.shape[0])
for i,x_i in enumerate(inputs):
    attention_scores_2[i]=torch.dot(x_i,input_query)
print(attention_scores_2)
```

```python
attention_weight_2=attention_scores_2/attention_scores_2.sum()
```

```python
attention_weight_2
```

```python
attention_weight_2.sum()
```

```python
def softmax_naive(x):
    return torch.exp(x)/torch.exp(x).sum(dim=0)
```

```python
softmax_naive(attention_scores_2)
```

```python
query=inputs[1]
context_vec_2=torch.zeros(query.shape)
for i,x_i in enumerate(inputs):
    context_vec_2+=attention_weight_2[i]*x_i
print(context_vec_2)
```

```python
attention_scores=torch.empty(6,6)
for i,x_i in enumerate(inputs):
    for j,x_j in enumerate(inputs):
        attention_scores[i,j]=torch.dot(x_i,x_j)
print(attention_scores)
```

```python
attention_scores=inputs@inputs.T
print(attention_scores)
```

```python
attention_weights=torch.softmax(attention_scores,dim=1)
print(attention_weights)
```

```python
attention_weights.sum(dim=1)
```

```python
all_context_vecs=attention_weights@inputs
print(all_context_vecs)
```
