```python
import re

class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self, text):

        preprocessed = re.split(r'([,.:;?_!"()\']|\s)', text)

        preprocessed = [
            item.strip()
            for item in preprocessed
            if item.strip()
        ]
        ids = [self.str_to_int.get(token, self.str_to_int["<UNK>"])for token in preprocessed]

        return ids

    def decode(self, ids):

        text = " ".join([self.int_to_str[i] for i in ids])

        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)

        return text
```

```python
from torch.utils.data import Dataset,DataLoader
class GPTDatasetV1(Dataset):
    def __init__(self,txt,tokenizer,max_length,stride):
        self.input_ids=[]
        self.target_ids=[]
        token_ids=tokenizer.encode(txt,allowed_special={"<|endoftext|>"})
        for i in range(0,len(token_ids)-max_length,stride):
            input_chunk=token_ids[i:i+max_length]
            target_chunk=token_ids[i+1:i+max_length+1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))
    def __len__(self):
        return len(self.input_ids)
    def __getitem__(self,idx):
        return self.input_ids[idx],self.target_ids[idx]

```

```python
def create_dataloader_v1(txt,batch_size=8,max_length=256,stride=64,shuffle=True,drop_last=True,num_workers=2):
    tokenizer=tiktoken.get_encoding("gpt2")
    dataset=GPTDatasetV1(txt,tokenizer,max_length,stride)
    dataloader=DataLoader(dataset,batch_size=batch_size,shuffle=shuffle,drop_last=drop_last,num_workers=num_workers)
    return dataloader
```

```python
import torch
import tiktoken

tokenizer = tiktoken.get_encoding("gpt2")

vocab_size = tokenizer.n_vocab      # 50257
output_dim = 768                    # embedding dimension

token_embedding_layer = torch.nn.Embedding(
    vocab_size,
    output_dim
)

context_length = 256
pos_embedding_layer = torch.nn.Embedding(
    context_length,
    output_dim
)
```

```python
data_iter = iter(dataloader)
inputs, targets = next(data_iter)

print(inputs.shape)
```

```python
token_embeddings = token_embedding_layer(inputs)

print(token_embeddings.shape)
```

```python
seq_len = inputs.shape[1]

pos_ids = torch.arange(seq_len)

pos_embeddings = pos_embedding_layer(pos_ids)

print(pos_embeddings.shape)
```

```python
input_embeddings = token_embeddings + pos_embeddings

print(input_embeddings.shape)
```
